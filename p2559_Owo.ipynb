### Seasonal Mean 2025

```python
# import xarray as xr
import numpy as np
import healpy as hp
import os

# Year
year = 2025

# Input and output directories
base_path = f"{PROJ2559_DIR}/ocean3d_dy/tco2559_proj/"
file_template = "0001_o3d_avg_wo_healpix_yearly_{year}_{month:02d}.grb"
output_dir = f"{PROJ2559_DIR}/ocean3d_dy/ana_var_wvl/"
os.makedirs(output_dir, exist_ok=True)
```

```python
# Define Seasons

# Seasons and months
seasons = {
    "DJF": [12, 1, 2],
    "MAM": [3, 4, 5],
    "JJA": [6, 7, 8],
    "SON": [9, 10, 11]
}
```

```python
# Load Monthly Datasets for the Season

import dask

def load_season_data(year, season_months):
    datasets = []

    for month in season_months:
        # Handle December of previous year for DJF
        file_year = year if month != 12 else year - 1
        file_path = os.path.join(base_path, file_template.format(year=file_year, month=month))

        print(f"Loading file: {file_path}")
        ds = xr.open_dataset(file_path, engine="cfgrib", chunks={"time": 30})

        # Extract the 3D variable: (time, oceanModel, values)
        datasets.append(ds["avg_wo"])

    # Concatenate over time
    print(f"Concatenating data for season...")
    season_data = xr.concat(datasets, dim="time")

    return season_data
```

```python
# Compute Seasonal Mean for All Vertical Levels

def compute_seasonal_mean(season_data):
    print("Computing seasonal mean for all vertical levels...")

    # Compute mean using Dask (lazy evaluation)
    seasonal_mean = season_data.mean(dim="time").compute()

    print("Seasonal mean computed!")
    print(f"Resulting shape: {seasonal_mean.shape}")
    return seasonal_mean
```

```python
# Step 5: Save seasonal mean to NetCDF with metadata.

def save_seasonal_mean_to_netcdf(seasonal_mean, original_data, year, season_name, output_dir):
    import os
    
    print(f"Saving seasonal mean to NetCDF: {season_name} {year}...")

    # Prepare the output filename
    filename = f"seasonal_mean_avg_wo_{season_name}_{year}.nc"
    output_path = os.path.join(output_dir, filename)

    # Prepare dataset for saving
    ds_out = xr.Dataset(
        {"avg_wo": (["oceanModel", "values"], seasonal_mean.data)},
        coords={
            "oceanModel": original_data["oceanModel"],
            "latitude": ("values", original_data["latitude"].values),
            "longitude": ("values", original_data["longitude"].values),
            "values": original_data["values"],
        },
        attrs=original_data.attrs  # Copy global attributes
    )

    # Also copy attributes for the variable
    ds_out["avg_wo"].attrs = original_data["avg_wo"].attrs

    # Save to file
    ds_out.to_netcdf(output_path)
    print(f"Saved: {output_path}")
```

```python
# Step 6: Interpolate selected vertical level (e.g., 210â€¯m) to 0.5Â° global lat-lon grid

import healpy as hp
import numpy as np

# Select one level for visualization, e.g., oceanModel index 30 (adjust index if needed)
level_idx = 30  # 210m in your case (if index 30 corresponds to nz1=31)
data_interp = seasonal_mean[level_idx, :].values

# Define 0.5Â° regular global lat-lon grid
lats = np.arange(-89.75, 90, 0.5)
lons = np.arange(-179.75, 180, 0.5)
grid_lon, grid_lat = np.meshgrid(lons, lats)

# Prepare output grid
grid_vvel = np.full(grid_lon.shape, np.nan)

print("Interpolating to 0.5Â° lat-lon grid...")

# Interpolate from HEALPix to regular grid
for i in range(grid_lon.shape[0]):
    for j in range(grid_lon.shape[1]):
        try:
            grid_vvel[i, j] = hp.get_interp_val(
                data_interp, grid_lon[i, j], grid_lat[i, j],
                lonlat=True, nest=True  # Use nest=True if your data is NESTED
            )
        except Exception:
            grid_vvel[i, j] = np.nan

print("Interpolation complete.")
```
