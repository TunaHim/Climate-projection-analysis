# Table of Contents : Ocean MLD

### Why this notebook matters

This notebook demonstrates 3D ocean mixed-layer-depth diagnostics from a storm-resolving climate projection, deriving MLD via a density-threshold algorithm and mapping its early-vs-late-period trend.

**Purpose** - Evaluate the mixed-layer depth (MLD) in a 2020-2039 storm-resolving coupled projection: load monthly 3D ocean temperature data, compute MLD with a density threshold, and produce time-series and spatial maps of MLD change over the projection period.

**Required input data** - 3D ocean potential temperature (`thetao`) NetCDF files on the HEALPix grid under `{PROJ2559_DIR}` (originally on DKRZ Levante; not included in this repository).

**Expected outputs** - MLD maps and time-series figures characterising projected MLD changes.

## Key techniques demonstrated

- Dask-backed 3D NetCDF/GRIB processing
- Density-threshold MLD algorithm (scipy)
- HEALPix handling with healpy
- Cartopy + cmocean mapping
- Early-vs-late-period trend comparison

## Trend: Seasonally means

### Calculate

```python
# Calculate for each seasons, the mean: for all the years. 

import os
import xarray as xr
import numpy as np
import pandas as pd

# === Configuration ===
input_dir = f"{PROJ2559_DIR}/ocean2d_dy/"
output_dir = f"{PROJ2559_DIR}/ocean2d_dy/ana_var_mld/"
os.makedirs(output_dir, exist_ok=True)

# === Define seasons ===
seasons = {
    "DJF": [12, 1, 2],
    "MAM": [3, 4, 5],
    "JJA": [6, 7, 8],
    "SON": [9, 10, 11],
}

# === Processing loop ===
for year in range(2020, 2040):
    print(f"\nProcessing year: {year}")

    try:
        # --- Load current year ---
        file_path = os.path.join(input_dir, f"0001_o2d_avg_mlotst125_healpix_daily_{year}.grb")
        ds = xr.open_dataset(file_path, engine="cfgrib")

        # --- Check if we need Dec from previous year for DJF ---
        if year > 2020:
            try:
                dec_file = os.path.join(input_dir, f"0001_o2d_avg_mlotst125_healpix_daily_{year-1}.grb")
                ds_prev = xr.open_dataset(dec_file, engine="cfgrib")
                ds_dec = ds_prev.sel(time=ds_prev.time.dt.month == 12)
                ds = xr.concat([ds_dec, ds], dim='time')
                ds_prev.close()
            except Exception as e:
                print(f"  Could not load Dec of previous year: {e}")

        # Loop over seasons
        for season, months in seasons.items():
            # Create boolean mask for months in season
            month_mask = ds.time.dt.month.isin(months)

            # For DJF: also ensure correct year labeling (e.g., Jan/Feb of current year with Dec of previous)
            if season == "DJF":
                # Use Jan-Feb of current year and Dec of previous year
                season_year = year
                mask = (
                    ((ds.time.dt.month == 12) & (ds.time.dt.year == year - 1)) |
                    ((ds.time.dt.month.isin([1, 2])) & (ds.time.dt.year == year))
                )
            else:
                season_year = year
                mask = (ds.time.dt.month.isin(months)) & (ds.time.dt.year == year)

            # Select data
            ds_season = ds.sel(time=mask)

            if ds_season.time.size == 0:
                print(f"  No data found for {season} {season_year}")
                continue

            # Compute seasonal mean
            seasonal_mean = ds_season['avg_mlotst125'].mean(dim='time', skipna=True)

            # Convert to dataset
            seasonal_ds = seasonal_mean.to_dataset(name='mld')
            seasonal_ds.attrs.update({
                'title': f'Seasonal Mean MLD ({season} {season_year})',
                'season': season,
                'year': season_year,
                'method': 'Seasonal mean of daily avg_mlotst125',
                'units': 'm',
                'long_name': 'Mixed layer depth defined by sigma theta 0.125 kg m-3'
            })

            # Output path
            out_file = os.path.join(output_dir, f"mld_{season.lower()}_{season_year}.nc")
            seasonal_ds.to_netcdf(out_file, encoding={'mld': {'zlib': True, 'complevel': 5}})
            print(f"  âœ” Saved {season} â†’ {out_file}")

            seasonal_ds.close()

        ds.close()

    except Exception as e:
        print(f"  âš ï¸ Error processing {year}: {str(e)}")
```

```python
# Seasonal mean calculation

import os
import numpy as np
import xarray as xr
from scipy.stats import linregress

# === Set season and directories ===
season = "son"  # Change to "mam", "jja", or "son"
data_dir = f"{PROJ2559_DIR}/ocean2d_dy/ana_var_mld/"
years = range(2020, 2040)
```

```python
# Load seasonal MLD data and stack

# === Load and stack all seasonal files ===
seasonal_ds_list = []

for year in years:
    file_path = os.path.join(data_dir, f"mld_{season}_{year}.nc")
    ds = xr.open_dataset(file_path)
    
    # Add time coordinate explicitly with nanosecond precision
    ds = ds.assign_coords(time=[np.datetime64(f"{year}-01-01T00:00:00", 'ns')])
    seasonal_ds_list.append(ds)

# Concatenate across time
seasonal_mld = xr.concat(seasonal_ds_list, dim="time")
print(seasonal_mld)
```

```python
# Compute pixel-wise trend

# === Prepare for trend calculation ===
mld_data = seasonal_mld.mld.transpose('time', 'values').values
trend_values = np.full(mld_data.shape[1], np.nan)

# === Linear regression per pixel ===
for i in range(mld_data.shape[1]):
    y = mld_data[:, i]
    valid = ~np.isnan(y)
    if np.sum(valid) >= 2:
        x = np.arange(len(y))[valid]
        slope, _, _, _, _ = linregress(x, y[valid])
        trend_values[i] = slope

# === Create output trend map ===
mld_trend = xr.DataArray(
    trend_values,
    dims=["values"],
    coords={"values": seasonal_mld["values"]},
    attrs={
        "units": "m/year",
        "season": season.upper(),
        "period": f"{years[0]}â€“{years[-1]}",
        "description": "Linear trend of MLD seasonal means"
    }
)
```

```python
# Save trend to NetCDF

output_path = os.path.join(data_dir, f"mld_trend_{season}_{years[0]}_{years[-1]}.nc")
mld_trend.to_dataset(name="mld_trend").to_netcdf(
    output_path,
    encoding={"mld_trend": {"zlib": True, "complevel": 5}}
)

print(f"Saved trend map for {season.upper()} to:\n{output_path}")
```

```python
# Compute the trend & save

# Sign still like the original

import os
import numpy as np
import xarray as xr
from scipy.stats import linregress

# === Setup ===
data_dir = f"{PROJ2559_DIR}/ocean2d_dy/ana_var_mld/"
years = range(2020, 2040)

# === Load yearly datasets ===
yearly_ds_list = []
for year in years:
    file_path = os.path.join(data_dir, f"mld_yearly_mean_{year}.nc")
    ds = xr.open_dataset(file_path)
    
    # Ensure proper time coordinate
    ds = ds.assign_coords(time=[np.datetime64(f"{year}-01-01T00:00:00", 'ns')])
    yearly_ds_list.append(ds)

# === Combine over time ===
yearly_mld = xr.concat(yearly_ds_list, dim="time")

# === Extract MLD values ===
mld_data = yearly_mld["mld"].transpose("time", "values").values
trend_values = np.full(mld_data.shape[1], np.nan)

# === Perform linear regression per pixel ===
for i in range(mld_data.shape[1]):
    y = mld_data[:, i]
    valid = ~np.isnan(y)
    if np.sum(valid) >= 2:
        x = np.arange(len(y))[valid]
        slope, _, _, _, _ = linregress(x, y[valid])
        trend_values[i] = slope

# === Create output DataArray ===
mld_trend_yearly = xr.DataArray(
    trend_values,
    dims=["values"],
    coords={"values": yearly_mld["values"]},
    attrs={
        "units": "m/year",
        "period": f"{years[0]}â€“{years[-1]}",
        "description": "Linear trend of yearly mean MLD"
    }
)

# === Save to NetCDF ===
output_path = os.path.join(data_dir, f"mld_trend_yearly_{years[0]}_{years[-1]}.nc")
mld_trend_yearly.to_dataset(name="mld_trend").to_netcdf(
    output_path,
    encoding={"mld_trend": {"zlib": True, "complevel": 5}}
)

print(f"âœ… Saved yearly MLD trend to:\n{output_path}")
```

```python
# Plotting

# Here the sign is flipped as our original data is in negative for MLD.

import os
import numpy as np
import xarray as xr
import healpy as hp
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import cmocean

# === File path ===
trend_file = f"{PROJ2559_DIR}/ocean2d_dy/ana_var_mld/mld_trend_yearly_2020_2039.nc"

# === Load trend data ===
ds = xr.open_dataset(trend_file)
trend_raw = ds.mld_trend.values

# === Define regular grid ===
lon = np.linspace(-180, 180, 720)
lat = np.linspace(-90, 90, 360)
grid_lon, grid_lat = np.meshgrid(lon, lat)
grid_trend = np.full_like(grid_lat, np.nan, dtype=np.float32)

# === Flip sign to make positive = deepening ===
grid_trend = -grid_trend

# === Interpolate HEALPix to regular lat-lon ===
for i in range(grid_lat.shape[0]):
    for j in range(grid_lat.shape[1]):
        try:
            val = hp.get_interp_val(trend_raw, grid_lon[i, j], grid_lat[i, j], lonlat=True, nest=True)
            if np.isfinite(val):
                grid_trend[i, j] = val
        except Exception:
            pass

# === Plot ===
fig = plt.figure(figsize=(14, 7))
ax = plt.subplot(1, 1, 1, projection=ccrs.Robinson(central_longitude=0))

# Choose color scale
vmin, vmax = -20, 20
levels = np.linspace(vmin, vmax, 21)

cf = ax.contourf(
    grid_lon,
    grid_lat,
    grid_trend,
    levels=levels,
    transform=ccrs.PlateCarree(),
    cmap="RdBu_r",
    extend="both"
)

# Map features
ax.set_title("Yearly Mean MLD Trend (2020â€“2039)", fontsize=14)
ax.coastlines(resolution='110m', linewidth=0.6)
ax.gridlines(draw_labels=False, linestyle="--", alpha=0.3)
ax.set_global()

# Colorbar
cbar = fig.colorbar(cf, ax=ax, orientation="horizontal", pad=0.05, fraction=0.04)
cbar.set_label("MLD Trend (m/year): Positive = Deepening, Negative = Shoaling", fontsize=11)

plt.tight_layout()
plt.show()
```

```python
import os
import xarray as xr
import numpy as np

input_dir = f"{PROJ2559_DIR}/ocean2d_dy/ana_var_mld/"
output_dir = input_dir  # Save to same folder

def compute_mld_5yr_mean(start_year, end_year, tag):
    years = range(start_year, end_year + 1)
    ds_list = []

    for yr in years:
        path = f"{input_dir}/mld_yearly_mean_{yr}.nc"
        ds = xr.open_dataset(path)
        ds = ds.expand_dims(year=[yr])  # Create a new 'year' dimension
        ds_list.append(ds)

    # Concatenate over year dimension
    ds_all = xr.concat(ds_list, dim='year')

    # Compute mean across years
    mean_5yr = ds_all['mld'].mean(dim='year', keep_attrs=True)
    mean_ds = mean_5yr.to_dataset(name='mld')
    mean_ds.attrs = {
        'description': f'5-year mean MLD ({start_year}-{end_year})',
        'units': 'm'
    }

    # Save
    out_path = os.path.join(output_dir, f"mld_5year_mean_{tag}.nc")
    mean_ds.to_netcdf(
        out_path,
        encoding={'mld': {'zlib': True, 'complevel': 5, 'dtype': 'float32'}}
    )
    print(f"âœ… Saved {tag} mean to {out_path}")
    return mean_5yr
```

```python
# Run for Baseline and Future

baseline = compute_mld_5yr_mean(2020, 2024, "baseline")
future = compute_mld_5yr_mean(2035, 2039, "future")
```

```python
# Calculate and Save the Difference

# Difference: future - baseline
mld_diff = future - baseline
mld_diff_ds = mld_diff.to_dataset(name='mld_diff')
mld_diff_ds.attrs = {
    'description': 'Difference of 5-year mean MLD (2035â€“2039 minus 2020â€“2024)',
    'units': 'm'
}

diff_path = os.path.join(output_dir, "mld_diff_2035-2039_minus_2020-2024.nc")
mld_diff_ds.to_netcdf(
    diff_path,
    encoding={'mld_diff': {'zlib': True, 'complevel': 5, 'dtype': 'float32'}}
)
print(f"âœ… Saved difference file to {diff_path}")
```

```python
import os
import numpy as np
import xarray as xr
import healpy as hp
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import cmocean

# === File Paths ===
data_dir = f"{PROJ2559_DIR}/ocean2d_dy/ana_var_mld/"
files = {
    "2020â€“2024 Mean": "mld_5year_mean_baseline.nc",
    "2035â€“2039 Mean": "mld_5year_mean_future.nc"
}

# === Set up 0.5Â° Regular Grid ===
lon = np.linspace(-180, 180, 720)
lat = np.linspace(-90, 90, 360)
grid_lon, grid_lat = np.meshgrid(lon, lat)

# === Interpolate and Store Data ===
mld_grid_dict = {}

for label, fname in files.items():
    ds = xr.open_dataset(os.path.join(data_dir, fname))
    mld_raw = -1 * ds['mld'].values  # Flip sign to positive depth
    grid_mld = np.full_like(grid_lat, np.nan, dtype=np.float32)

    print(f"Interpolating {label}...")

    for i in range(grid_lat.shape[0]):
        for j in range(grid_lat.shape[1]):
            try:
                val = hp.get_interp_val(
                    mld_raw, grid_lon[i, j], grid_lat[i, j],
                    lonlat=True, nest=True
                )
                if np.isfinite(val):
                    grid_mld[i, j] = val
            except Exception:
                pass

    mld_grid_dict[label] = np.clip(grid_mld, 0.1, 1000)

```

```python
# === Plot Side-by-Side ===
fig, axs = plt.subplots(1, 2, figsize=(16, 7),
                        subplot_kw={'projection': ccrs.Robinson(central_longitude=0)})

levels = np.linspace(0, 500, 21)

for ax, (label, grid_mld) in zip(axs, mld_grid_dict.items()):
    cf = ax.contourf(
        grid_lon, grid_lat, grid_mld,
        levels=levels,
        transform=ccrs.PlateCarree(),
        cmap=cmocean.cm.deep,
        extend='max'
    )
    ax.set_title(f"{label}", fontsize=13)
    ax.coastlines(resolution='110m', linewidth=0.6)
    ax.gridlines(draw_labels=False, linestyle='--', color='gray', alpha=0.3)
    ax.set_global()

# === Adjust spacing BEFORE adding colorbar ===
plt.subplots_adjust(left=0.05, right=0.95, top=0.88, bottom=0.12, wspace=0.05)

# === Add Shared Colorbar ===
cbar = fig.colorbar(cf, ax=axs, orientation='horizontal', pad=0.05, aspect=50)
cbar.set_label("Mixed Layer Depth (m)", fontsize=12)

# === Title ===
plt.suptitle("Mixed Layer Depth (0.5Â° Grid): 5-Year Means", fontsize=16)
plt.show()
```

```python
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import cmocean

# === Load Data ===
data_path = f"{PROJ2559_DIR}/ocean2d_dy/ana_var_mld/mld_yearly_interpolated_2020_2039.nc"
ds = xr.open_dataset(data_path)

years = ds.year.values
lat = ds.lat.values
lon = ds.lon.values
grid_lon, grid_lat = np.meshgrid(lon, lat)

# Convert to positive MLD values
mld_data = -1 * ds.mld.values  # shape: (time, lat, lon)

# === Define Zoom Region (50Â°Nâ€“70Â°N, âˆ’60Â°E to âˆ’20Â°E) ===
lat_min, lat_max = 50, 70
lon_min, lon_max = -60, -20

lat_mask = (lat >= lat_min) & (lat <= lat_max)
lon_mask = (lon >= lon_min) & (lon <= lon_max)

lat_zoom = lat[lat_mask]
lon_zoom = lon[lon_mask]
mld_zoom = mld_data[:, lat_mask, :][:, :, lon_mask]
grid_lon_zoom, grid_lat_zoom = np.meshgrid(lon_zoom, lat_zoom)
```

```python

# === Plot Setup ===
fig, ax = plt.subplots(figsize=(9, 6), subplot_kw={'projection': ccrs.PlateCarree()})
ax.set_extent([lon_min, lon_max, lat_min, lat_max], crs=ccrs.PlateCarree())
ax.coastlines(resolution='10m')
ax.add_feature(cfeature.LAND, facecolor='lightgray', zorder=1)
ax.gridlines(draw_labels=True, linestyle='--', color='gray', alpha=0.3)

vmin, vmax = 0, 500  # Adjust if needed
cmap = cmocean.cm.deep

# Initial frame
mesh = ax.pcolormesh(grid_lon_zoom, grid_lat_zoom, mld_zoom[0],
                     transform=ccrs.PlateCarree(), cmap=cmap, vmin=vmin, vmax=vmax)
title = ax.set_title(f"MLD (m), Year: {years[0]}", fontsize=14)

# Colorbar
cbar = fig.colorbar(mesh, ax=ax, orientation='horizontal', shrink=0.7, pad=0.05)
cbar.set_label("Mixed Layer Depth (m)")

# === Animation Update Function ===
def update(frame):
    mesh.set_array(mld_zoom[frame].ravel())
    title.set_text(f"MLD (m), Year: {years[frame]}")
    return [mesh, title]

# === Create Animation ===
anim = FuncAnimation(fig, update, frames=len(years), interval=600, blit=False)

plt.close()
HTML(anim.to_jshtml())
```
